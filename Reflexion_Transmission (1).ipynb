# Coefficients de réflexion et de transmission

Un phénomène physique ondulatoire peut transporter de la puissance. La puissance par unité de surface transportée s'écrit toujours comme le produit de deux grandeurs :
$$ J = a.b$$
Par exemple :
1. **câble électrique :** $a=$ tension, $b=$ courant,
1. **onde sonore :** $a=$ pression acoustique, $b=$ vitesse de déplacement des molécules,
1. **onde lumineuse :** $a=$ champ électrque $E$, $b=$ champ magnétique $H$.

L'impédance est une quantité définie par $\underline{Z} = \frac{\underline{a}}{\underline{b}}$. Si ces deux ondes sont de simples ondes planes propagatives $\propto e^{j\left(\omega t - q x\right)}$, alors $\underline{Z}$ est un nombre complexe **constant**, caractéristique des propriétés du milieu :
$$\underline{Z} = \pm \alpha \frac{\underline{q}}{\omega}$$ ($+$ pour une onde progressive, $-$ pour une onde régressive).
On parle dans ce cas d'impédance **spécifique** du milieu.

Quand les ondes ne sont pas simplement propagatives mais partiellement stationnaires (lorsqu'il y a superposition de différentes ondes incidentes et réfléchies par exemple), on peut montrer que $\underline{Z}$ ne dépend jamais du temps, mais peut alors dépendre de la position $x$ dans le milieu. 

La puissance transportée par l'onde, dite **puissance active**, est égale à la puissance moyenne $\left<J\right>$ qui est proportionnelle à la partie réelle de l'impédance.

La partie imaginaire de $\underline{Z}$ renseigne sur la puissance **réactive**, qui ne se propage pas mais n'est que temporairement stockée dans le milieu.

À l'interface entre deux milieux, il y a rupture des propriétés des matériaux (densité, compressibilité, capacité...) et donc d'impédance spécifique. Il peut alors y avoir des phénomènes de réflexion (et de transmision) de l'onde à cette interface.

Le coefficient de réflexion **en amplitude** pour l'onde $b$ **à l'interface** entre deux milieux d'impédances **spécifiques** $\underline{Z}_1$ et $\underline{Z}_2$ est donné par :
$$ \underline{r} = \frac{\underline{Z}_1-\underline{Z}_2}{\underline{Z}_1+\underline{Z}_2} $$ 

Le programme ci-dessous simule le comportement à l'interface des ondes réfléchies et transmises en fonction des impédances $\underline{Z}_1$ et $\underline{Z}_2$ :

```python
import numpy as np
import matplotlib.pylab as plt
from matplotlib import animation,rc
from IPython.display import HTML

def OP(q,x,A,o,t):
    return A*np.exp(1j*(o*t-q*x))
    
fig = plt.figure()

ax = plt.axes(xlim=(-5,5),ylim=(-12,8))

#################### Animation

N = 11
lines = [plt.plot([], [])[0] for _ in range(N)] #lines to animate


def init():
    
    #init lines
    for line in lines:
        line.set_data([], [])

    return lines #return everything that must be updated

def animate(i,Z1,Z2):
    #animate lines
    dt = 0.011
    
    c=1

    r = (Z1-Z2)/(Z1+Z2)
    t = r+1.0
    
    x1 = np.linspace(-5.0,0.0,512)
    x2 = np.linspace(0.0,5.0,512)
    
    Ai = 1.0
    Ar = r*Ai
    At = t*Ai
    off = 2*max(np.abs(Ai),np.abs(Ar),np.abs(At))
    
    qi = 2*np.pi/1.0
    o=qi*c
    qr = -qi
    qt = qi*Z2/Z1
    
    oi = OP(qi,x1,Ai,o,i*dt)
    ore = OP(qr,x1,Ar,o,i*dt)
    ot = OP(qt,x2,At,o,i*dt)
    o1 = oi+ore
    pt = ot.real*(Z2*ot).real
    p1 = o1.real*(Z1*(oi-ore)).real
    
 #   txt1 = "$Z_1=${z1}".format(z1=Z1)
 #   txt2 = "$Z_2=${z2}".format(z2=Z2)
 #   plt.text(-4.5,1.8*off,txt1,fontsize=16)
 #   plt.text(0.5,1.8*off,txt2,fontsize=16)
    
    lines[0].set_data(x1,np.real(o1))
    lines[1].set_data(x2,np.real(ot))
    lines[2].set_data(x2,np.real(ot+off))
    lines[3].set_data(x1,np.real(oi+off))
    lines[4].set_data(x1,np.real(ore+off))
    lines[5].set_data([-5.0,5.0],[off,off])
    lines[6].set_data([-5.0,5.0],[0,0])
    lines[7].set_data([0.0,0.0],[-4*off,2*off])
    lines[8].set_data(x1,p1-2*off)
    lines[9].set_data(x2,pt-2*off)
    lines[10].set_data([-5.0,5.0],[-2*off,-2*off])
    
    return lines #return everything that must be updated

plt.close()
```

```python
Z1 = 2.0
Z2 =-1j*0.5
```

## Lancement de l'animation

Le programme crée une vidéo qui présente la propagation :
1. en haut de la figure : des ondes incidente et réfléchie dans le milieu 1, et de l'onde transmise dans le milieu 2
1. au milieu de la figure : de l'onde totale incidente + réfléchie dans le milieu 1, et de l'onde transmise dans le milieu 2
1. en bas de la figure : de la puissance instanée

Cliquez dans la cellule ci-dessous et validez par [SHIFT]+[ENTREE]. Attention, le rendu peut prendre un peu de temps.

Remarquez la continuité des ondes totale (milieu 1) et transmise (milieu 2) à l'interface, ce qui illustre que la fréquence est la même pour toutes les ondes (ce qui n'est pas le cas des longueurs d'onde).

```python
anim = animation.FuncAnimation(fig, animate, init_func=init,
                                frames=300, fargs=(Z1,Z2), interval=20, blit=False)
#HTML(anim.to_html5_video())
plt.close()
rc('animation', html='jshtml')
anim
```
